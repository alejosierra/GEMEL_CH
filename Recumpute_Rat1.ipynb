```python
from utils import calc_r_at_1
import json
from pathlib import Path
import numpy as np
```

```python
inference_folder = Path("./inference_results/melart/")
for seed in inference_folder.iterdir():
    if seed.is_dir():
        res_file=seed / "results.json"
        if res_file.exists():
            with open(res_file, "r") as f:
                results = json.load(f)
            predictions = [p["pred"] for p in results["predictions"]]
            targets = [p["target"] for p in results["predictions"]]
            r_at_1 = calc_r_at_1(predictions, targets)
            prev_metric = results["metrics"].get("r_at_1", 0.0)
            print(f"Seed: {seed.name}, Previous R@1: {prev_metric:.4f} %, Recomputed R@1: {r_at_1:.4f} %")
            results["metrics"]["r_at_1"] = r_at_1
            with open(res_file, "w") as f:
                json.dump(results, f, indent=4)
```

```python
# compute the average and standard deviation of R@1 across all seeds from the results

r_at_1_values = []
for seed in inference_folder.iterdir():
    if seed.is_dir():
        res_file=seed / "results.json"
        if res_file.exists():
            with open(res_file, "r") as f:
                results = json.load(f)
            r_at_1_values.append(results["metrics"]["r_at_1"])

print(f"Average R@1: {np.mean(r_at_1_values):.4f} %")
print(f"Standard Deviation R@1: {np.std(r_at_1_values):.4f} %")
```
